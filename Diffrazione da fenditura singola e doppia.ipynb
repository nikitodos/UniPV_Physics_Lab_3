##Introduzione

La diffrazione è un fenomeno di interferenza, che si verifica quando un'onda incontra un ostacolo o una fenditura. Si distingue tra diffrazione di Fresnel e diffrazione di Fraunhofer, a seconda che le distanze tra sorgente, schermo e apertura siano più o meno grandi; quella di Fraunhofer, nello specifico, è quella che coinvolge grandi distanze, ottenibili mediante lenti.

La figura di diffrazione prodotta da una fenditura rettilinea si può costruire dal principio di Huygens-Fresnel-Kirchhoff, e differisce dalla propagazione per raggi rettilinei.

1.png

Dalla costruzione, che fa uso di una $f(\vartheta)$ detta fattore di obliquità, si trova che i minimi della figura di diffrazione per una fenditura rettilinea singola corrispondono per:

$$b \ \sin \vartheta = \lambda, 2 \lambda, \ldots = p \ \lambda, \quad p=1,2,\ldots$$

e sono accompagnati da diversi massimi, a loro volta più o meno intensi.

Se si dispone di 2 fenditure rettilinee identiche e regolarmente distanziate, da una costruzione simile alla precedente (che fa uso delle proprietà della serie geometrica) si trova che i minimi della figura da queste prodotte rispondono alla relazione:

$$d \ \sin \vartheta = \lambda / 2 , 3 \lambda /2, \ldots = (m + \ ^1/_2) \ \lambda, \quad m=1,2,\ldots$$

Che deriva da una relazione ben più generale, ovvero:

$$I(\vartheta)=I_0 \Big(\frac{\sin \alpha}{\alpha}\Big)^2 \Big(\frac{\sin (N \beta)}{\beta}\Big)^2, \quad \alpha = \frac{\pi b \sin \vartheta}{\lambda}, \quad \beta = \frac{\pi d \sin \vartheta}{\lambda}$$

Con $N$ numero di fenditure.

L'effetto è dunque tanto maggiore quanto più le dimensioni dell'ostacolo/apertura diventano confrontabili con la lunghezza d'onda della radiazione utilizzata, e i punti di minimo sono accompagnati da punti di massimo primari o secondari di intensità più o meno elevata.

Dalla legge precedente, inoltre, risulta evidente che l'intensità della figura di interferenza è modulata dalla diffrazione; i massimi principali saranno massimi di diffrazione, mentre i massimi secondari saranno massimi di interferenza.

## Procedura sperimentale

Montato l'apparato sperimentale sul modello della figura precedente, si procede al campionamento delle distanze dei minimi rispetto al centro della figura di diffrazione prodotta sullo schermo, che può essere costituito da un foglio di carta millimetrata.

In questo caso particolare, una buona strategia può essere quella di fotografare lo schermo per misurare le distanze, oppure misurare la distanza picco-picco direttamente dallo schermo, tenendo conto di tale approssimazione nell'analisi successiva.

Ottenuti i valori di $p$ e $\vartheta = x / r \simeq \sin \vartheta$, dove $x$ e $r$ sono rispettivamente la distanza di un buio di diffrazione dal centro della figura e la distanza tra il LASER e lo schermo, fissa, si può sfruttare la relazione dei minimi della figura di diffrazione per fenditura singola vista nell'introduzione per ricavare la larghezza delle fenditure utilizzate $b$.

Ripetendo lo stesso campionamento con le fenditure doppie, minimizzando la $I(\vartheta)$ si ottiene la seconda relazione vista nell'introduzione, relativa ai minimi della figura di interferenza, da cui si possono ricavare i valori di $b$ e $d$. In questo caso, però, la grandezza utilizzata è $\eta = y / r \simeq \sin \eta$, con $y$ distanza di un buio di interferenza dal centro della figura sullo schermo.

## Discussione dei risultati e conclusioni

Operando con angoli molto piccoli, il seno è ben approssimato dal valore dell'angolo in radianti, e questo è pressochè identico al rapporto delle distanze trasversali tra le frange e la distanza reticolo-schermo. Inoltre, grazie all'utilizzo di fotografie ad alta risoluzione dello schermo graduato è stato possibile ottenere misurazioni molto precise anche degli ordini di minimi relativi alle fenditure più larghe, e dunque meno osservabili (a parità di distanza sorgente-schermo).

Tale accorgimento, combinatamente all'utilizzo di una sorgente ad alta lunghezza di coerenza, come il LASER, ha permesso di effettuare stime pressoché perfette dei parametri di passo e larghezza delle fenditure, come evidenziato sul foglio di calcolo.

Nelle schede di regressione lineare, inoltre, si possono osservare tre dei valori calcolati in modo precedente, ricavati attraverso la regressione lineare del programma di calcolo e graficati, che forniscono ulteriore prova della correttezza dei risultati trovati.

Pertanto, si stabilisce che sono raggiunti con successo tutti gli obiettivi della prova.
