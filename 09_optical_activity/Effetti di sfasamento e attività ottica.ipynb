##Introduzione

Il fenomeno dell’attività ottica consiste nella rotazione della direzione di
polarizzazione lineare di un’onda elettromagnetica mentre si propaga all’interno
di un mezzo.
La figura mostra un esempio di mezzo destrogiro: la direzione di polarizzazione
ruota verso destra o in senso orario, visto dalla direzione dell’osservatore. Per un
mezzo levogiro, la direzione di polarizzazione ruota invece in senso antiorario.

image.png

L’attività ottica è descritta dalla seguente legge:

$$\alpha = K \cdot L$$

dove $α$ è l’angolo di rotazione (positivo per un mezzo destrogiro, negativo per
un mezzo levogiro), $L$ è la lunghezza di propagazione nel materiale e $K$ è detto potere rotatorio (legato alla birifrangenza circolare, ossia alla differenza di indice di rifrazione per le due onde polarizzate circolarmente).

In una soluzione, il potere rotatorio è proporzionale alla concentrazione $c$ di
sostanza otticamente attiva e con quest'ultima si ottiene la legge di Biot:

$$\alpha = k \cdot c \cdot L$$

dove $k$ è detto potere rotatorio specifico, ha le dimensioni di deg·cm$^2$/g e dipende dalla lunghezza d’onda λ secondo la legge:

$$k(\lambda) = {A \over k^2}$$

##Scopo dell'esperienza

Lo scopo degli esperimenti è di:

- Verificare la legge dell’attività ottica e la dipendenza del potere rotatorio specifico dalla lunghezza d’onda;
- Misurare il potere rotatorio specifico di soluzioni di saccarosio, fruttosio e glucosio alla lunghezza d’onda di riferimento $λ=589$ nm;
- Studiare i fenomeni di mutarotazione del glucosio e della inversione della
soluzione di saccarosio.

## Strumentazione

Per raggiungere gli obiettivi della prova si dispone di:

- 1 polarimetro con LED alle lunghezze d’onda $λ=630$ nm (rosso), $580$ nm (giallo), $525$ nm (verde), $468$ nm (blu);
- 1 cilindro graduato per polarimetro;
- 1 becher e 1 cilindro graduato con base;
- Saccarosio, fruttosio, glucosio;
- Soluzione di acido cloridrico al 25%;
- Carta stagnola (per coprire la soluzione di HCl quando non utilizzata).

Avendo a che fare con composti chimici potenzialmente pericolosi, prima di operare è necessario tenere a mente alcuni accorgimenti. ## Procedura sperimentale

La prova si articola in 4 esperimenti, illustrati di seguito.

### Esperimento 1: verifica della legge dell’attività ottica

Sciolta una quantità di saccarosio (fino a $30-40$ g) in $80$ ml di acqua, si osserva che l’angolo di rotazione è proporzionale all’altezza della soluzione. Dunque, sciolte diverse quantità di saccarosio nella stessa quantità di acqua, si misura il potere rotatorio, verificando che è proporzionale alla concentrazione, e da questo si determina il potere rotatorio specifico.

Osservazione: dato che utilizzando un LED verde o blu l’angolo di rotazione è maggiore, è preferibile utilizzare uno di questi con il polarizzatore.

### Esperimento 2: potere rotatorio del saccarosio, fruttosio, glucosio

Determinata la sezione del cilindro graduato di riferimento, si scioglie una certa quantità di zucchero in acqua, versando la soluzione nel cilindro
graduato. Si misura dunque l’angolo di rotazione alle lunghezze d’onda dei quattro LED.

Da queste misure si verifica che la dispersione del potere rotatorio specifico è del tipo $k(\lambda)$, illustrato in precedenza, mediante il fit della costante A; da quest'ultimo passaggio è possibile determinare il potere rotatorio specifico a $λ=589$ nm con relativa incertezza, tenendo presente che, dai valori tabulati a $λ=589$ nm, $c = 1$ g/mL, $T=20°$ C:

- il saccorosio e il glucosio sono destrogiri, con $k=6.65$ e $5.24$ deg cm$^2$/g;
- il fruttosio è levogiro, con $k=−9.12$ deg cm$^2$/g.

### Esperimento 3: mutarotazione del glucosio

Si prepara una soluzione di glucosio usando acqua calda. A lunghezza d’onda fissata, si misura l’angolo e il potere rotatorio in funzione del tempo.

Il D-glucosio o destrosio si presenta in due forme isomeriche:
- α-D-glucosio, con $k=11.2$ deg cm$^2$/g per $\lambda = 589$ nm
- β-D-glucosio, con $k=1.87$ deg cm$^2$/g per $\lambda = 589$ nm

In soluzione acquosa avviene il fenomeno della mutarotazione: le due forme si trasformano l'una nell’altra, fino a raggiungere le proporzioni di equilibrio
36% α-D-glucosio / 64% β-D-glucosio, che forniscono $k=(0.36·11.2+0.64·1.87)=5.23$ deg cm$^2$/g per $\lambda = 589$ nm.

### Esperimento 4: inversione della soluzione di saccarosio

Preparata una soluzione di saccarosio usando acqua calda, si aggiunge qualche ml di soluzione ad HCl preparata in precedenza. Così, fissata una lunghezza d'onda, si misurano l'angolo $𝛼$ e il potere rotatorio in funzione del tempo.

In soluzione acquosa, il saccarosio si scinde in glucosio e fruttosio per idrolisi, ma la reazione avviene spontaneamente solo ad alta temperatura; la presenza di acido aumenta notevolmente la velocità di reazione.

A reazione completa, la soluzione di zucchero invertito ha un potere rotatorio negativo risultante dalla miscela di glucosio e fruttosio, e vale $k = {1 \over 2} (k_\text{glucosio} + k_\text{fruttosio}) = −1.94$ deg cm$^2$/g. Il problema che si presenta in quest'ultima prova è che la reazione richiede molte ore.

## Discussione dei risultati e conclusioni

Anziché preparare diverse soluzioni a concentrazione controllata si osserva che $V=S·L$, dove $S$ è la sezione del cilindro graduato, e quindi la legge di Biot diventa $α=k·P/S$, con $S$ calcolata una sola volta, dividendo il volume del cilindro con la sua lunghezza.

Dal primo esperimento si comprende che, a sezione fissa, l’angolo di rotazione non dipende dalla quantità di acqua, ma solo dal peso della sostanza disciolta, ossia dal numero di molecole otticamente attive. Inoltre, dal fit dei dati sperimentali si ottiene un andamento quasi perfettamente rettilineo, con una trascurabile quota iniziale di regressione lineare.

Dal secondo esperimento si ottiene un buon accordo con i valori di k relativi al saccarosio; minore l'accordo con i restanti composti, in particolare con il glucosio, quasi certamente dovuto all'effetto di mutarotazione del glucosio (che inizialmente è in prevalenza $\alpha$-D-glucosio).

Dal terzo esperimento si osserva che effettivamente il glucosio raggiunge un valore di $k$ per $\lambda = 589$ piuttosto simile a quello tabulato; il valore segnalato dai dati sperimentali si è mantenuto pressoché costante durante il resto della prova, per cui non è stato riportato.

Dalla quarta prova, infine, si verifica l'inversione del saccarosio per catalisi con l'acido cloridrico, ma la reazione è troppo lunga per poter essere analizzata fino in fondo nei tempi disponibili in laboratorio.

Per i risultati raggiunti, si stabilisce che sono raggiunti con successo tutti gli obiettivi della prova.
