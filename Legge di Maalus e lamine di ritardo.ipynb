##Introduzione

La polarizzazione della luce indica la direzione dell'oscillazione del vettore campo elettrico (ortogonale a quello campo magnetico) durante la propagazione dell'onda nello spazio-tempo. Si distinguono, in particolare:

- Polarizzazione lineare, in cui il campo elettrico oscilla lungo una direzione, nel piano perpendicolare alla direzione di propagazione;
- Polarizzazione circolare, in cui il campo elettrico ruota nel piano perpendicolare alla direzione di propagazione.

image.png

Un polarizzatore lineare produce luce polarizzata linearmente, con il campo che oscilla lungo l’asse ottico del polarizzatore. Può essere realizzato con il
polaroid, oppure con materiali birifrangenti (prisma di Nicol), oppure
utilizzando la riflessione all’angolo di Brewster (tale per cui se un'onda incide su una superficie proprio a $\theta _{B}$, si trova che l'onda riflessa è polarizzata perpendicolarmente al piano di propagazione).

Una lamina $𝜆/2$ introduce uno sfasamento di $𝜋$ fra le due componenti dei campi, trasformando un’onda elettromagnetica polarizzata linearmente in un’altra, con direzione di polarizzazione ottenuta per riflessione speculare rispetto all’asse ottico della lamina.

Una lamina $𝜆/4$, invece, introduce uno sfasamento di $𝜋/2$ tra le due componenti dei campi. Essa dunque trasforma un’onda elettromagnetica polarizzata linearmente in sé stessa, se la direzione di polarizzazione è a 0° o 90° con l’asse ottico, oppure un’onda polarizzata circolarmente, se la direzione di polarizzazione è a 45° con uno degli assi ottici della lamina, oppure un’onda polarizzata ellitticamente, se l’angolo è diverso da 0°, 45°, 90° o 135°. Al contrario, una lamina λ/4 trasforma un’onda polarizzata circolarmente in un’onda polarizzata linearmente lungo uno degli assi ottici. 

##Scopo dell'esperienza

Gli obiettivi dell'esperienza sono:

- produrre polarizzazione lineare, per verificare la legge di Malus $I(\vartheta) = I_0 \cos^2 \vartheta$ ;
- produrre luce polarizzata linearmente su lamine $λ/2$ e $λ/4$;
distinguere luce polarizzata linearmente/circolarmente/ellitticamente
- distinguere tra luce polarizzata e luce non polarizzata.

## Discussione dei risultati e conclusioni

Dal primo esperimento è evidente l'accordo tra dati sperimentali e legge di Maalus, in base ai valori di regressione lineare risultanti.

Anche dal secondo esperimento si riscontra accordo con la teoria per lamine mezz'onda, come risulta dai valori tabulati.

Dal terzo esperimento, si verifica che effettivamente la lamina quarto d'onda non modifica l'intensità del fascio luminoso; tuttavia:
- Quando la lamina è orientata ad un angolo $\vartheta_1 = 0°, 90°$ rispetto al primo polarizzatore, la luce uscente è polarizzata linearmente, per cui mettendo un secondo polarizzatore lineare ad un angolo $\vartheta_2$ l’intensità uscente varia come $\cos^2 \vartheta_2$.
- Quando la lamina è orientata ad un angolo $\vartheta_1 = 45°, 135°$ rispetto al primo polarizzatore, la luce uscente è polarizzata circolarmente, per cui mettendo un secondo polarizzatore lineare l’intensità uscente non dipende dalla direzione di quest’ultimo.
- Quando la lamina è orientata ad un angolo $\vartheta_1 \neq 0°, 45°, 90°, 135°$ rispetto al primo polarizzatore, la luce uscente è polarizzata ellitticamente, per cui mettendo un secondo polarizzatore lineare l’intensità uscente varia da $𝐼_0 \cos^2 \vartheta_2$ a $𝐼_0 \sin^2 \vartheta_2$.

Dal quarto esperimento, infine, risulta verificata la natura circolare della polarizzazione del LED, in seguito all'azione della lamina quarto d'onda orientata come descritto nel procedimento. Inoltre, i valori di $I$ ottenuti permettono di concludere che la luce del LED è intrinsecamente non polarizzata, come previsto.

Pertanto, si stabilisce che sono raggiunti con successo tutti gli obiettivi della prova.
